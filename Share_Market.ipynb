```python
outlier_columns=['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']
```

```python
def outlier(df,column):
    q1= df[column].quantile(0.25)
    q3= df[column].quantile(0.75)

    iqr= q3-q1

    upper_threshold= q3 + (1.5*iqr)
    lower_threshold= q1 - (1.5*iqr)

    df[column]= df[column].clip(lower_threshold, upper_threshold)
```

```python
# Outlier Handling
for i in outlier_columns:
    outlier(df_2,i)
```

```python
def select_model(df, algorithm):
    # x, y spliting
    x= df.drop(columns=["Close", "Date", "Company", "Ticker"])
    y= df["Close"]

    # Get the Train data size
    train_size = int(0.8 * df.shape[0])
    test_size = int(df.shape[0] - train_size)
    test_size,train_size

    # Data Spliting with temporal order
    X_train = x[:train_size]
    X_test = x[train_size:]
    y_train = y[:train_size]
    y_test = y[train_size:]

    model= algorithm().fit(X_train,y_train)
    y_pred_train= model.predict(X_train)
    y_pred_test= model.predict(X_test)

    #Checking the accuracy score
    train_mse= mean_squared_error(y_train, y_pred_train)
    train_rmse= np.sqrt(train_mse)
    train_mae= mean_absolute_error(y_train, y_pred_train)
    train_r2= r2_score(y_train, y_pred_train)

    test_mse= mean_squared_error(y_test, y_pred_test)
    test_rmse= np.sqrt(test_mse)
    test_mae= mean_absolute_error(y_test, y_pred_test)
    test_r2= r2_score(y_test, y_pred_test)

    metrics_train={"Data": "Train Data",
        "Name":algorithm.__name__,
        "R2_score":train_r2,
        "Mean_squared_error":train_mse,
        "Root_mean_squared_error":train_rmse,
        "Mean_absolute_error":train_mae
        }

    metrics_test={"Data": "Test Data",
           "Name":algorithm.__name__,
           "R2_score":test_r2,
           "Mean_squared_error":test_mse,
           "Root_mean_squared_error":test_rmse,
           "Mean_absolute_error":test_mae
              }

    print(metrics_train)
    print(metrics_test)
    print("")
```

```python
y_test_data_lR = select_model(df_3, LinearRegression)
y_test_data_DT = select_model(df_3, DecisionTreeRegressor)
y_test_data_RF = select_model(df_3, RandomForestRegressor)
y_test_data_AB = select_model(df_3, AdaBoostRegressor)
y_test_data_ETR = select_model(df_3, ExtraTreesRegressor)
y_test_data_GBR = select_model(df_3, GradientBoostingRegressor)
y_test_data_XGB = select_model(df_3, XGBRegressor)
y_test_data_LGB = select_model(df_3, lgb.LGBMRegressor)
```

```python
def ploting_data_candlestick(df, column):
    # Filter the dataframe based on the column
    df_company = df[df[column] == 1].reset_index(drop=True)

    # Create a candlestick chart
    fig = go.Figure(data=[go.Candlestick(x=df_company['Date'],
                                         open=df_company['Open'],
                                         high=df_company['High'],
                                         low=df_company['Low'],
                                         close=df_company['Close'])])

    # Add a title to the chart
    fig.update_layout(
        title=f"Candlestick Chart for {column}",
        xaxis_rangeslider_visible=False
    )

    # Display the chart
    fig.show()
```

```python
def ploting_data_scatter(df, column):

    df_company = df[df[column] == 1].reset_index(drop = True)

    fig = go.Figure(go.Scatter(x=df_company["Date"],y=df_company["Close"],mode='lines',name=f'{column} Close'))
    fig.update_layout(
        xaxis=dict(
            rangeselector=dict(
                buttons=list([
                    dict(count= 1,label ="All", step = "all", stepmode= "backward"),
                    dict(count= 5,label="5 Years",step="year",stepmode="backward"),
                    dict(count= 1,label="1 Year",step="year",stepmode="backward"),
                    dict(count= 6,label="6 Months",step="month",stepmode="backward"),
                    dict(count= 1,label="1 Month",step="month",stepmode="backward"),
                    dict(count= 7,label="1 Week",step="day",stepmode="backward"),
                    dict(count= 1,label="1 Day",step="day",stepmode="backward")
                ])
            )
        )
    )
    # Add a title to the chart
    fig.update_layout(
        title=f"Scatter Plot for {column}",
        xaxis_rangeslider_visible=False
    )

    fig.show()
```

```python
import plotly.graph_objects as go
def actual_predicted_comparison(df_org, df_test, column):
    x_test_org = df_org.iloc[18444:][df_org[column] == 1]
    x_test_pred_company = df_test[df_test[column] == 1]


    fig = go.Figure()

    # Add the first scatter plot (e.g., Company Close prices Original)
    fig.add_trace(go.Scatter(
        x=x_test_org["Date"],
        y=x_test_org["Close"],
        mode='lines',
        name=f'{column} Close',
        line=dict(color='blue')
    ))

    # Add the second scatter plot (e.g., Company Close prices Predicted)
    fig.add_trace(go.Scatter(
        x=x_test_org["Date"],
        y=x_test_pred_company["y_pred"],
        mode='lines',
        name=f'{column} Close',
        line=dict(color='orange')
    ))

    # Update layout with the range selector
    fig.update_layout(
        xaxis=dict(
            rangeselector=dict(
                buttons=list([
                    dict(count=1, label="All", step="all", stepmode="backward"),
                    dict(count=5, label="5 Years", step="year", stepmode="backward"),
                    dict(count=1, label="1 Year", step="year", stepmode="backward"),
                    dict(count=6, label="6 Months", step="month", stepmode="backward"),
                    dict(count=1, label="1 Month", step="month", stepmode="backward"),
                    dict(count=7, label="1 Week", step="day", stepmode="backward"),
                    dict(count=1, label="1 Day", step="day", stepmode="backward")
                ])
            )
        ),
        title=f"{column} Stock Data",
        xaxis_title="Date",
        yaxis_title="Stock Price"
    )

    # Show the plot
    fig.show()
```
